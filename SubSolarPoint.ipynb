```python
def findSubSolarLat(InputTime):
    
    TropicLat = 23.43585   # degrees extent of the Tropic of Cancer and Capricorn
    
    WinterSolsticeTime = datetime(2026, 6, 21, 18, 24)   # 21 June 2026 at 18:24
    SummerSolsticeTime = datetime(2026, 12, 22, 6, 50)   # 22 December 2026 at 06:50
    TimeBetweenSolstices = SummerSolsticeTime - WinterSolsticeTime
    
    TimeSinceSolstice    = InputTime - WinterSolsticeTime
    FractionToSolstice   = TimeSinceSolstice / TimeBetweenSolstices
    DegreesSinceSolstice = FractionToSolstice * 180

    SubSolarLat = TropicLat * (np.cos(np.deg2rad(DegreesSinceSolstice)))

    return SubSolarLat
```

```python
TropicLat = 23.43585   # degrees extent of the Tropic of Cancer and Capricorn
KmPerLat = 40000 / 360  # how many kilometres in a degree of latitude
```

```python
# retrieve times
Now = datetime.now()
JustBeforeNow = Now - timedelta(seconds=0.5)
JustAfterNow  = Now + timedelta(seconds=0.5)

# retrieve sub solar points
SubSolarLatNow = findSubSolarLat(Now)
SubSolarLatJustBeforeNow = findSubSolarLat(JustBeforeNow)
SubSolarLatJustAfterNow = findSubSolarLat(JustAfterNow)

# calculate rate of movement
SubSolarDelta      = -1 * (SubSolarLatJustAfterNow - SubSolarLatJustBeforeNow)
SubSolarDeltaDay   = SubSolarDelta * 86400
SubSolarDeltaKm    = SubSolarDelta  * KmPerLat
SubSolarDeltaKmDay = SubSolarDeltaDay  * KmPerLat
SubSolarDeltaCm    = SubSolarDeltaKm * 1000 * 100

SubSolarHoursPerKm = (1/SubSolarDeltaKmDay) * 24

# calcualte location of the subsolar point relative to its solstice location
SubSolarRiseNow = (TropicLat - SubSolarLatNow) * KmPerLat # [km] 

# read out the findings
print(f'Estimated Current Subsolar Point Info:')
print(f'{SubSolarLatNow.round(5)}°N')
print(f'{SubSolarRiseNow.round(3)} km south of June solstice')
print(f'{SubSolarDeltaCm.round(3)} cm/s or {SubSolarDeltaKmDay.round(3)} km/day ({SubSolarDeltaDay.round(3)}°/day) Southward')
print(f'1 km travelled every {SubSolarHoursPerKm.round(2)} hours')
```

```python
# Generate 100000 evenly spaced values from 0 to 360 degrees
degrees = np.linspace(0, 360, 100000)

# Convert to radians and take the sine
sines = np.sin(np.radians(degrees))

# Plot histogram
fig, ax = plt.subplots(figsize=(10, 6))
ax.hist(sines, bins=100, color='steelblue', edgecolor='black', linewidth=0.5)
ax.set_xlabel('sin(θ)', fontsize=12)
ax.set_ylabel('Count', fontsize=12)
ax.set_title('Histogram of sin(θ) for θ ∈ [0°, 360°]', fontsize=14, fontweight='bold')
# ax.set_yscale('log')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
