```python
import numpy as np
from matplotlib import pyplot as plt

import pandas as pd # for opening csv files

from pathlib import Path      # used to play with pathnames to save 

from scipy.interpolate import interp1d # use for interpolation

import matplotlib.dates as mdates # for adding date labels to plots

from matplotlib.colors import hsv_to_rgb, LinearSegmentedColormap

# import xarray as xr

# import zipfile as zp          # used for unzipping ppi files
# from pathlib import Path      # used to play with pathnames to save 
# from datetime import datetime # used to manipulate time :)


# import wradlib as wr          # used for having fun with radar data

# from PIL import Image         # used for creating gif loops
# import os                     # used for retrieving file names
```

```python
# import the csv files into a large python data array

# create an empty array for storing the mean wind in the lowest layer
LowMeanWind = np.full([367], np.nan) #o ne more day than the leap year 2024 since the first entry (day 0) is Nan


# empty arrays to store all of the wind data at every height for every day (5000 is a maximum possible number of vertical points, but most will remain nans)
FullYearWindAlts = np.full([367,5000], np.nan)  # altitude at which the wind measurements are taken

FullYearWindSpeeds = np.full([367,5000], np.nan)
FullYearWindDirs = np.full([367,5000], np.nan)
FullYearWindU = np.full([367,5000], np.nan)
FullYearWindV = np.full([367,5000], np.nan)


MonthLengths = [31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31] # number of days in each month of the year

Station = 'Townsville'

if (Station == 'Townsville'):
    StationID = 95282

MaxLowHeight = 5000 # [m] top height of where to consider the mean wind of the lowest layer

year = 2024

for monthi in range(1,13): # months 1 - 12 (Jan to Dec)
    for dayi in range(1,MonthLengths[monthi-1] + 1): # days of month 1 to 31 or whatever

        doyi = int(np.sum(MonthLengths[0:monthi-1]) + dayi)

        hour = 0
        
        YYYY = str(year).zfill(4)
        MM = str(monthi).zfill(2)
        DD = str(dayi).zfill(2)
        
        hh = str(hour).zfill(2)
        
        SoundingsFolder = '/home/563/sg3241/TownsvilleSoundings/'
        SoundingsFile = YYYY + MM + DD + hh + '-' + str(StationID) + '.csv'
        
        SoundingPath = SoundingsFolder + SoundingsFile



        if not Path(SoundingPath).is_file():
            print('No file found for ' + SoundingPath)
        else:
            WorkingSounding = pd.read_csv(SoundingPath, usecols=['geopotential height_m', 'wind direction_degree', 'wind speed_m/s'])

            WorkingNumAlts = np.shape(WorkingSounding['geopotential height_m'])[0]

            # rad = np.deg2rad(WorkingSounding['wind direction_degree'])
            # WorkingSounding['u'] = -WorkingSounding['wind speed_m/s'] * np.sin(rad)   # eastward
            # WorkingSounding['v'] = -WorkingSounding['wind speed_m/s'] * np.cos(rad)   # northward

        
            # # calculate the mean wind (CAUTION mean of the data points, not weighted by depth) in the lowest ZZZZ metres
            # LowMeanWind[doyi] = WorkingSounding[WorkingSounding['geopotential height_m'] < MaxLowHeight]['u'].mean()

            FullYearWindAlts[doyi, 0:WorkingNumAlts] = WorkingSounding['geopotential height_m']
            
            FullYearWindSpeeds[doyi, 0:WorkingNumAlts] = WorkingSounding['wind speed_m/s']
            FullYearWindDirs[doyi, 0:WorkingNumAlts] = WorkingSounding['wind direction_degree']
            FullYearWindU[doyi, 0:WorkingNumAlts] = -WorkingSounding['wind speed_m/s'] * np.sin( np.deg2rad(WorkingSounding['wind direction_degree']) )   # eastward
            FullYearWindV[doyi, 0:WorkingNumAlts] = -WorkingSounding['wind speed_m/s'] * np.cos( np.deg2rad(WorkingSounding['wind direction_degree']) )   # northward
```

```python
# Create a date range for all days in 2024
ProtoFullYearWindDates = pd.date_range(start='2024-01-01', end='2024-12-31', freq='D', tz=None)

# Convert to numpy datetime64 array
ProtoFullYearWindDatesArray = ProtoFullYearWindDates.values

# Create the final 367-element array with NaT at index 0
FullYearWindDates = np.empty(367, dtype='datetime64[ns]')
FullYearWindDates[0] = np.datetime64('NaT')
FullYearWindDates[1:] = ProtoFullYearWindDatesArray
```

```python
# CREATE COLOUR BAR FOR WIND DIRECTION

# Define the 4 cardinal colours with smooth transitions
# Order: N(red) -> E(yellow) -> S(blue) -> W(green) -> N(red)
cardinal_colors = [
    (1.0, 0.0, 0.0),  # North   - Red
    (1.0, 1.0, 0.0),  # East    - Yellow
    (0.0, 0.0, 1.0),  # South   - Blue
    (0.5, 0.0, 1.0),  # West    - Violet
    (1.0, 0.0, 0.0),  # North   - Red (repeated to close the cycle)
]

cardinal_cmap = LinearSegmentedColormap.from_list('cardinal_dirs', cardinal_colors, N=256)
```

```python
# CHAD CREATED PLOT OF WINDSPEEDS THROUGHOUT THE YEAR

import warnings
warnings.filterwarnings('ignore')

# Step 0: Use original dates and data as-is
WindDates = FullYearWindDates
FullYearWindSpeeds_clean = FullYearWindSpeeds
FullYearWindDirs_clean = FullYearWindDirs
FullYearWindAlts_clean = FullYearWindAlts

# Step 1: Check altitude values
valid_alts = FullYearWindAlts_clean[~np.isnan(FullYearWindAlts_clean)]
if len(valid_alts) > 0:
    min_alt = np.nanmin(valid_alts)
    max_alt = np.nanmax(valid_alts)
    if min_alt < 0:
        FullYearWindAlts_clean = np.abs(FullYearWindAlts_clean)
        min_alt = np.nanmin(FullYearWindAlts_clean[~np.isnan(FullYearWindAlts_clean)])
        max_alt = np.nanmax(FullYearWindAlts_clean[~np.isnan(FullYearWindAlts_clean)])
else:
    raise ValueError("No valid altitude data")

# ADJUSTABLE HEIGHT LIMIT (in km)
height_limit_km = 20
height_limit_m = height_limit_km * 1000

# ADJUSTABLE SPEED LIMITS
speed_limit_plot = 60       # m/s — colour axis limit for wind speed plot
speed_limit_saturation = 20 # m/s — speed at which dual plot reaches full saturation

# Step 2: Create a common altitude grid (multiples of 10 m starting at 0)
common_alt_grid = np.arange(0, min(max_alt, height_limit_m) + 10, 10)

# Step 3: Create interpolated grids for each day
interpolated_wind_speeds = []
interpolated_wind_dirs = []
valid_days = []

for day in range(FullYearWindSpeeds_clean.shape[0]):
    alts = FullYearWindAlts_clean[day, :]
    speeds = FullYearWindSpeeds_clean[day, :]
    dirs = FullYearWindDirs_clean[day, :]
    
    valid_mask = ~(np.isnan(alts) | np.isnan(speeds) | np.isnan(dirs))
    alts_valid = alts[valid_mask]
    speeds_valid = speeds[valid_mask]
    dirs_valid = dirs[valid_mask]
    
    if len(alts_valid) > 1:
        sort_idx = np.argsort(alts_valid)
        alts_valid = alts_valid[sort_idx]
        speeds_valid = speeds_valid[sort_idx]
        dirs_valid = dirs_valid[sort_idx]
        
        try:
            f_speed = interp1d(alts_valid, speeds_valid, kind='linear', 
                              bounds_error=False, fill_value=np.nan)
            speed_interp = f_speed(common_alt_grid)
            
            f_dir = interp1d(alts_valid, dirs_valid, kind='linear', 
                            bounds_error=False, fill_value=np.nan)
            dir_interp = f_dir(common_alt_grid)
            
            min_alt_day = np.min(alts_valid)
            max_alt_day = np.max(alts_valid)
            outside_range = (common_alt_grid < min_alt_day) | (common_alt_grid > max_alt_day)
            speed_interp[outside_range] = np.nan
            dir_interp[outside_range] = np.nan
            
            interpolated_wind_speeds.append(speed_interp)
            interpolated_wind_dirs.append(dir_interp)
            valid_days.append(day)
        except Exception as e:
            print(f"Warning: Could not interpolate day {day}: {e}")
            continue

# Step 4: Create 2D arrays for plotting
speed_grid = np.full((len(common_alt_grid), len(WindDates)), np.nan)
dir_grid = np.full((len(common_alt_grid), len(WindDates)), np.nan)

for i, day in enumerate(valid_days):
    speed_grid[:, day] = interpolated_wind_speeds[i]
    dir_grid[:, day] = interpolated_wind_dirs[i]

# Convert altitude grid to km for plotting
common_alt_grid_km = common_alt_grid / 1000.0

# -----------------------------------------------------------------------
# Create rotated HSV colormap: East=red, West=cyan
# -----------------------------------------------------------------------
n_colors = 256
hues = (np.linspace(0, 1, n_colors) - 0.25) % 1.0
hsv_colors = np.stack([hues, np.ones(n_colors), np.ones(n_colors)], axis=1)
rgb_colors = hsv_to_rgb(hsv_colors.reshape(1, -1, 3)).reshape(-1, 3)
rotated_hsv_cmap = LinearSegmentedColormap.from_list('rotated_hsv', rgb_colors, N=n_colors)

# -----------------------------------------------------------------------
# Helper: mid-month tick positions and labels
# -----------------------------------------------------------------------
def get_mid_month_ticks(year=2024):
    import pandas as pd
    mid_dates = []
    labels = []
    month_names = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
                   'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
    for month in range(1, 13):
        # Middle of each month (day 15)
        mid = pd.Timestamp(year=year, month=month, day=15)
        mid_dates.append(mid)
        labels.append(month_names[month - 1])
    return mid_dates, labels

# -----------------------------------------------------------------------
# Helper: apply dark grid styling
# -----------------------------------------------------------------------
def apply_dark_grid(ax):
    ax.set_facecolor('black')

    # --- X axis: thick lines every 3 months (month boundaries) ---
    ax.xaxis.set_major_locator(mdates.MonthLocator(bymonth=[1, 4, 7, 10]))
    ax.xaxis.set_major_formatter(plt.NullFormatter())  # No labels on major ticks
    ax.grid(which='major', axis='x', color='white', linewidth=1.2, alpha=0.7)

    # --- X axis: thin lines every 1 month ---
    ax.xaxis.set_minor_locator(mdates.MonthLocator())
    ax.xaxis.set_minor_formatter(plt.NullFormatter())
    ax.grid(which='minor', axis='x', color='white', linewidth=0.3, alpha=0.7)

    # --- X axis: labels at mid-month ---
    mid_dates, labels = get_mid_month_ticks(year=2024)
    ax2_x = ax.twiny()
    ax2_x.set_xlim(ax.get_xlim())
    ax2_x.xaxis.set_ticks_position('bottom')
    ax2_x.xaxis.set_label_position('bottom')
    ax2_x.spines['bottom'].set_position(('outward', 20))
    ax2_x.set_xticks(mid_dates)
    ax2_x.set_xticklabels(labels, color='white', fontsize=9)
    ax2_x.tick_params(axis='x', which='both', length=0, colors='white')
    for spine in ax2_x.spines.values():
        spine.set_visible(False)

    # --- Y axis: thick lines every 5 km ---
    ax.yaxis.set_major_locator(plt.MultipleLocator(5))
    ax.grid(which='major', axis='y', color='white', linewidth=1.2, alpha=0.7)

    # --- Y axis: thin lines every 1 km ---
    ax.yaxis.set_minor_locator(plt.MultipleLocator(1))
    ax.grid(which='minor', axis='y', color='white', linewidth=0.3, alpha=0.3)

    ax.tick_params(colors='white', which='both')
    ax.xaxis.label.set_color('white')
    ax.yaxis.label.set_color('white')
    ax.title.set_color('white')
    for spine in ax.spines.values():
        spine.set_edgecolor('white')

# -----------------------------------------------------------------------
# Step 5: Create the plots
# -----------------------------------------------------------------------
plt.style.use('dark_background')
fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(20, 10))
fig.patch.set_facecolor('black')

# Plot 1: Wind Speed (capped at speed_limit_plot)
levels1 = np.linspace(0, speed_limit_plot, 21)
im1 = ax1.imshow(speed_grid[:, 1:], aspect='auto', origin='lower',
                  extent=[mdates.date2num(WindDates[1].astype('datetime64[ms]').astype('O')),
                          mdates.date2num(WindDates[-1].astype('datetime64[ms]').astype('O')),
                          0, height_limit_km],
                  cmap='gist_heat', vmin=0, vmax=speed_limit_plot,
                  interpolation='none')

ax1.set_ylabel('Altitude (km)', fontsize=12)
ax1.set_title('Wind Speed from ' + Station + ' Soundings Throughout 2024', fontsize=14, fontweight='bold')
cbar1 = plt.colorbar(im1, ax=ax1)
cbar1.set_label('Wind Speed (m/s)', fontsize=11, color='white')
cbar1.ax.yaxis.set_tick_params(color='white')
plt.setp(cbar1.ax.yaxis.get_ticklabels(), color='white')
apply_dark_grid(ax1)

im2 = ax2.imshow(dir_grid[:, 1:], aspect='auto', origin='lower',
                  extent=[mdates.date2num(WindDates[1].astype('datetime64[ms]').astype('O')),
                          mdates.date2num(WindDates[-1].astype('datetime64[ms]').astype('O')),
                          0, height_limit_km],
                  cmap=cardinal_cmap, vmin=0, vmax=360,
                  interpolation='none')

ax2.set_ylabel('Altitude (km)', fontsize=12)
ax2.set_title('Wind Direction from ' + Station + ' Soundings Throughout 2024', fontsize=14, fontweight='bold')
cbar2 = plt.colorbar(im2, ax=ax2, ticks=[0, 45, 90, 135, 180, 225, 270, 315, 360])
cbar2.ax.set_yticklabels(['N', 'NE', 'E', 'SE', 'S', 'SW', 'W', 'NW', 'N'], color='white')
cbar2.set_label('Wind Direction', fontsize=11, color='white')
cbar2.ax.yaxis.set_tick_params(color='white')
apply_dark_grid(ax2)

# Plot 3: Wind Direction (Hue) + Speed (Value/Brightness)
# Plot 3: Wind Direction (Hue) + Speed (Value/Brightness)
dir_plot = dir_grid[:, 1:]
spd_plot = speed_grid[:, 1:]

# Map direction to cardinal_cmap colours (same as Plot 2)
norm_dir = dir_plot / 360.0  # Normalise to [0, 1]
rgb_from_cmap = cardinal_cmap(norm_dir)[:, :, :3]  # Shape: (alt, time, 3) — drop alpha

# Scale brightness by wind speed
val_array = np.clip(spd_plot / speed_limit_saturation, 0, 1)

# Multiply RGB by brightness
rgb_array = rgb_from_cmap * val_array[:, :, np.newaxis]

# Handle NaNs — set to black
nan_mask = np.isnan(dir_plot) | np.isnan(spd_plot)
rgb_array[nan_mask, :] = 0.0

im3 = ax3.imshow(rgb_array, aspect='auto', origin='lower',
                  extent=[mdates.date2num(WindDates[1].astype('datetime64[ms]').astype('O')),
                          mdates.date2num(WindDates[-1].astype('datetime64[ms]').astype('O')),
                          0, height_limit_km],
                  interpolation='none')
ax3.xaxis_date()
ax3.set_xlabel('Date', fontsize=12, color='white')
ax3.set_ylabel('Altitude (km)', fontsize=12)
ax3.set_title(f'Wind Direction from {Station} Soundings Throughout 2024\nWith Speed Indicated by Brightness (Full colour at {speed_limit_saturation} m/s)',
              fontsize=14, fontweight='bold')

cbar3 = plt.colorbar(im2, ax=ax3, ticks=[0, 45, 90, 135, 180, 225, 270, 315, 360])
cbar3.ax.set_yticklabels(['N', 'NE', 'E', 'SE', 'S', 'SW', 'W', 'NW', 'N'], color='white')
cbar3.set_label('Wind Direction', fontsize=11, color='white')
cbar3.ax.yaxis.set_tick_params(color='white')
apply_dark_grid(ax3)

plt.tight_layout()

SaveFolder = '/scratch/v46/sg3241/tmp/pngImages/'
SaveFile   = Station + 'SoundingsWindPlots2024.png'
SavePath = SaveFolder + SaveFile

if not Path(SaveFolder).exists():
    print('Creating Folder: ' + SaveFolder)
    Path(SaveFolder).mkdir(parents=True, exist_ok=True)

plt.savefig(SavePath, bbox_inches='tight', facecolor='k', dpi = 300)
plt.close()
```
