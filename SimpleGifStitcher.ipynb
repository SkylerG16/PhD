```python
# NOTEBOOK NAME
# SimpleGifStitcher.ipynb
# NOTEBOOK NAME

from PIL import Image
```

```python
#  CHOOSE YOUR RADAR
RadarIDno = '22'  # 22 is Mackay, 106 is Townsville, 66 is Mt Staplyton (Brisbane) , 50 is Marabong (near Bris)

# CHOOSE YOUR DAY
# the day in consideration (YYYYMMDD) and time (hhmmss)       ALL IN UTC !!!
RadarYear  = 2024
RadarMonth = 2
RadarDay   = 14

# add leading zeros for strings
YYYY = str(RadarYear).zfill(4)
MM = str(RadarMonth).zfill(2)
DD = str(RadarDay).zfill(2)

# write out the data in one string
RadarFileDate  = YYYY + MM + DD

# CHOOSE YOUR QUALITY CONTROL SETTINGS
# # taken from Aragon et al. 2024
# MinValidZDR = -4 # ZDR DOESNT WORK FOR MACKAY EARLY 2024 BECAUSE OF THE SOURCE RADAR DATA IN STORAGE
# MaxValidZDR =  4
MinValidRhoHV = 0.85

VarName = 'Reflectivity'
```

```python
# TRY AND MAKE A STITCHED GIF
# COMPRESSED DATA TEST

gif_files = [
    "/scratch/v46/sg3241/tmp/gifImages/Horz/" + RadarIDno  + "/" + RadarFileDate + "/" + RadarIDno  + "_" + RadarFileDate + "_" + VarName + "_RhoHVlim" + str(MinValidRhoHV*100)[0:2] + "_2000m.gif",
    "/scratch/v46/sg3241/tmp/gifImages/Horz/" + RadarIDno  + "/" + RadarFileDate + "/" + RadarIDno  + "_" + RadarFileDate + "_" + VarName + "_RhoHVlim" + str(MinValidRhoHV*100)[0:2] + "_2000mCompressed.gif",
]

# Open GIFs
gifs = [Image.open(f) for f in gif_files]

# Check frame counts
n_frames = gifs[0].n_frames
for gif in gifs:
    if gif.n_frames != n_frames:
        raise ValueError("All GIFs must have the same number of frames")

frames = []

for frame_idx in range(n_frames):
    imgs = []

    for gif in gifs:
        gif.seek(frame_idx)
        imgs.append(gif.convert("RGBA"))

    w, h = imgs[0].size

    # Create 2x1 canvas
    combined = Image.new("RGBA", (2 * w, 1 * h))

    combined.paste(imgs[0], (0, 0))
    combined.paste(imgs[1], (w, 0))

    frames.append(combined)

# Use duration from first GIF
gifs[0].seek(0)
duration = gifs[0].info.get("duration", 100)

frames[0].save(
    "/scratch/v46/sg3241/tmp/gifImages/Stitched/" + RadarIDno  + "_2" + RadarFileDate + "_" + VarName + "_RhoHVlim" + str(MinValidRhoHV*100)[0:2] + "Compressed.gif",
    save_all=True,
    append_images=frames[1:],
    duration=duration,
    loop=0,
    disposal=2,
)
```
